# Question - 1

Q1. The weight (in kgs) of a new product launched by Company X is as follows:
[11, 12, 9, 8, 8, 11, 10, 10, 13, 10, 12, 14, 11, 12, 8, 9, 10, 9, 8, 10]
Calculate the confidence interval of the product weight.


```python
import pandas as pd
import numpy as np
import scipy.stats as st
from scipy.stats import sem
import matplotlib.pyplot as plt
```

```python
# create sample data
data = [11, 12, 9, 8, 8, 11, 10, 10, 13, 10, 12, 14, 11, 12, 8, 9, 10, 9, 8, 10]

print(data)
```

```python
# Code here

# 1. Set sample size and confidence level
n = len(data)
confidence = 0.95

# 2. Calculate mean and standard error of the mean (SEM)
mean = np.mean(data)
std_err = st.sem(data)  # Computes s / sqrt(n) automatically

# 4. Calculate the t-distribution confidence interval
lower_bound, upper_bound = st.t.interval(confidence, df=n-1, loc=mean, scale=std_err)

# 5. Output the results
print(f"Sample Mean: {mean:.2f} kg")
print(f"Standard Error (SEM): {std_err:.3f} kg")
print(f"95% Confidence Interval: [{lower_bound:.2f} kg, {upper_bound:.2f} kg]")
```

To calculate the 95% confidence interval for the product weight, we use a Student's t-distribution because the population standard deviation is unknown and the sample size is small (n = 20).



Step-by-Step Calculation

1. Find the Sample Size (n)
Count of data points: n = 20

2. Calculate the Sample Mean (x-bar)
x-bar = sum(x) / n = (11 + 12 + ... + 10) / 20 = 205 / 20 = 10.25 kg

3. Calculate the Sample Standard Deviation (s)
s = square_root( sum((x - x-bar)^2) / (n - 1) ) approx 1.743 kg

4. Find the Standard Error of the Mean (SEM)
SEM = s / square_root(n) = 1.743 / square_root(20) approx 0.390 kg

5. Find the Critical t-value (t*)
Degrees of freedom (df) = n - 1 = 19
For a 95% confidence level (alpha = 0.05), looking at a t-distribution table: t* = 2.093

6. Calculate the Margin of Error (ME)
ME = t* * SEM = 2.093 * 0.390 approx 0.816 kg

7. Determine the Confidence Interval
CI = x-bar +/- ME = 10.25 +/- 0.816
Lower Limit = 10.25 - 0.816 = 9.434 kg
Upper Limit = 10.25 + 0.816 = 11.066 kg


# Question - 2

Q2.Hospital ABC noted the average weight of the 12-year-old children who come to visit their pediatric ward for a specific type of stomach disease.
weight = [28, 26, 27, 28, 29, 26, 30, 26, 30, 26, 25, 25, 28, 27, 27, 29, 28, 30, 28, 29]
Check the hypothesis that the average weight of kids having this disease is 28 kgs.

```python
# Code here
import numpy as np
import scipy.stats as stats

# 1. Input the dataset
weights = [28, 26, 27, 28, 29, 26, 30, 26, 30, 26, 25, 25, 28, 27, 27, 29, 28, 30, 28, 29]

# 2. Define the hypothesized population mean
hypothesized_mean = 28

# 3. Perform the one-sample t-test
t_statistic, p_value = stats.ttest_1samp(weights, hypothesized_mean)

# 4. Calculate descriptive statistics for context
sample_mean = np.mean(weights)
sample_std = np.std(weights, ddof=1)  # Sample standard deviation (n-1)

# 5. Output results
print(f"Sample Mean: {sample_mean:.2f} kg")
print(f"Sample Standard Deviation: {sample_std:.3f} kg")
print(f"t-statistic: {t_statistic:.3f}")
print(f"p-value: {p_value:.4f}")

# 6. Interpret the results
alpha = 0.05
if p_value < alpha:
    print("Result: Reject the null hypothesis (Significant evidence that average weight is NOT 28 kg).")
else:
    print("Result: Fail to reject the null hypothesis (No significant evidence that average weight differs from 28 kg).")





```
