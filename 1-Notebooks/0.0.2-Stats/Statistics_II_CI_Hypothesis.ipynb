## 1. Confidence Interval

A confidence interval (CI) is a range of values that's likely to contain the population parameter with a certain level of confidence (e.g., 95%). The confidence interval is calculated differently based on whether we are dealing with a large sample size (Z-distribution) or a small sample size (T-distribution).

### Example 1: Z-Distribution

#### Business Scenario:
A company wants to estimate the average time customers spend on their website. From a random sample of 50 customers, they calculate the sample mean time as 100 minutes with a known population standard deviation of 15 minutes. They want to construct a 95% confidence interval to estimate the true average time customers spend on their website.




```python
import numpy as np
import scipy.stats as stats

# Business Scenario:
# A company wants to estimate the average time customers spend on their website.
# Sample mean time = 100 minutes, Population standard deviation = 15 minutes, Sample size = 50
# The company wants to construct a 95% confidence interval.

# Given data
sample_mean = 100
population_std_dev = 15
sample_size = 50
confidence_level = 0.95

# Calculate the confidence interval using scipy.stats.norm.interval
confidence_interval = stats.norm.interval(confidence_level, loc=sample_mean, scale=population_std_dev/np.sqrt(sample_size))

print(f"95% Confidence Interval: {confidence_interval}")
```

### Example 2: T-Distribution

A startup wants to estimate the average sales amount per transaction in a small market. They take a random sample of 15 transactions and calculate the sample mean sales amount as `$50` with a sample standard deviation of
 `$10`. Because the sample size is small, they will use the T-distribution to construct a 95% confidence interval for the true average sales amount per transaction.

```python
# Business Scenario:
# A startup wants to estimate the average sales amount per transaction in a small market.
# Sample mean sales amount = $50, Sample standard deviation = $10, Sample size = 15
# The startup wants to construct a 95% confidence interval.

# Given data
sample_mean = 50
sample_std_dev = 10
sample_size = 15
confidence_level = 0.95

# Calculate the confidence interval using scipy.stats.t.interval
confidence_interval = stats.t.interval(confidence_level, df=sample_size-1, loc=sample_mean, scale=sample_std_dev/np.sqrt(sample_size))

print(f"95% Confidence Interval: {confidence_interval}")
```

## 2. Hypothesis Testing

### Hypothesis Testing for Means for a Single Sample Z-test

A factory produces light bulbs with a guaranteed mean lifespan of 1000 hours. A quality control manager takes a random sample of 50 bulbs and finds the mean lifespan to be 1050 hours. The standard deviation is known to be 200 hours. The manager wants to test if this observed difference is statistically significant, using a Z-test at the 5% significance level.

```python
# Business Scenario:
# A factory produces light bulbs with a guaranteed mean lifespan of 1000 hours.
# A sample of 50 bulbs shows a mean lifespan of 1050 hours, with a standard deviation of 200 hours.
# The manager wants to test if this difference is statistically significant (5% significance level).

# Given data
sample_mean = 1050
population_mean = 1000
population_std_dev = 200
sample_size = 50
alpha = 0.05

# Z-test using scipy.stats.zscore (manual approach, as there's no direct z-test function)
z_value = (sample_mean - population_mean) / (population_std_dev / np.sqrt(sample_size))

# Calculate the p-value using the survival function (sf) for a two-tailed test
p_value = 2 * stats.norm.sf(abs(z_value))

print(f"Z-value: {z_value}")
print(f"P-value: {p_value}")
```

```python
# If the p-value is less than the significance level, then we can reject the null hypothesis and conclude that the difference in conversion rates is statistically significant
if p_value < alpha:
    print('The difference in conversion rates is statistically significant.')
else:
    print('The difference in conversion rates is not statistically significant.')
```

### 2. Hypothesis Testing for Proportion for a Single Sample Z-test

A smartphone manufacturer claims that 50% of its customers are satisfied with their product. A market research firm surveys 100 customers and finds that 40% of them are satisfied. The firm wants to test whether this observed proportion is significantly different from the manufacturer’s claim, using a Z-test at the 5% significance level.

```python
# Business Scenario:
# A smartphone manufacturer claims that 50% of its customers are satisfied.
# A survey of 100 customers shows that 40% are satisfied.
# The firm wants to test if this observed proportion is significantly different (5% significance level).

# Given data
sample_proportion = 0.4
population_proportion = 0.5
sample_size = 100
alpha =0.05
# Z-test for proportions calculation
z_value = (sample_proportion - population_proportion) / np.sqrt((population_proportion * (1 - population_proportion)) / sample_size)

# Calculate the p-value using the survival function (sf) for a two-tailed test
p_value = 2 * stats.norm.sf(abs(z_value))

print(f"Z-value: {z_value}")
print(f"P-value: {p_value}")
```

Proportion =	√( p̂(1 − p̂) / n )

```python
# If the p-value is less than the significance level, then we can reject the null hypothesis and conclude that the difference in conversion rates is statistically significant
if p_value < alpha:
    print('The difference in conversion rates is statistically significant.')
else:
    print('The difference in conversion rates is not statistically significant.')
```

### 3. Hypothesis Testing for Means for two-sample t-test

An e-commerce company wants to compare the average spending of customers who visited their website through two different marketing channels: social media and email campaigns. They collect data from two independent samples, where sample 1 consists of 10 customers who came via social media and sample 2 consists of 10 customers who came via email campaigns. They want to determine if there is a significant difference in the average spending between the two groups using a two-sample t-test.

```python
# Business Scenario:
# An e-commerce company wants to compare the average spending of customers from social media and email campaigns.
# Two independent samples of 10 customers each are taken from both groups.
# The company wants to determine if there's a significant difference in average spending between the groups.

# Given data for two samples
sample1 = np.array([85, 86, 88, 75, 78, 94, 98, 79, 71, 80])
sample2 = np.array([91, 92, 95, 96, 94, 90, 92, 93, 85, 88])
alpha = 0.05

# T-test calculation using scipy.stats.ttest_ind
t_value, p_value = stats.ttest_ind(sample1, sample2)

print(f"T-value: {t_value}")
print(f"P-value: {p_value}")
```

```python
# If the p-value is less than the significance level, then we can reject the null hypothesis and conclude that the difference in conversion rates is statistically significant
if p_value < alpha:
    print('The difference in conversion rates is statistically significant.')
else:
    print('The difference in conversion rates is not statistically significant.')
```

### 4. Hypothesis Testing for Proportion - Two Sample T-test (A/B testing)

A software company wants to determine whether a new feature increases user engagement. They conduct an A/B test where Group A (control group) uses the old version of the software, and Group B (test group) uses the new version with the feature. Group A consists of 200 users, with 100 of them engaging with the software, while Group B consists of 300 users, with 180 of them engaging. The company wants to know if the new feature significantly increases user engagement, using a t-test for the difference in proportions.

```python
import numpy as np
import scipy.stats as stats

# Business Scenario:
# Group A (control) has 200 users. Group B (treatment) has 300 users.
# Each user is either engaged (1) or not engaged (0).

# Simulating engagement data for Group A (control) and Group B (treatment)
np.random.seed(42)  # For reproducibility

# Example: 200 users in Group A, 100 engaged (50% engagement rate)
group_a_data = np.concatenate([np.ones(100), np.zeros(100)])

# Example: 300 users in Group B, 180 engaged (60% engagement rate)
group_b_data = np.concatenate([np.ones(180), np.zeros(120)])

# Conducting a two-sample t-test using scipy.stats.ttest_ind
t_value, p_value = stats.ttest_ind(group_a_data, group_b_data)

print(f"T-value: {t_value}")
print(f"P-value: {p_value}")
```

```python
# If the p-value is less than the significance level, then we can reject the null hypothesis and conclude that the difference in conversion rates is statistically significant
if p_value < alpha:
    print('The difference in conversion rates is statistically significant.')
else:
    print('The difference in conversion rates is not statistically significant.')
```
