## Q2: Implement Gradient Descent

Use the function $f(x) = x^2 -4x + 3$ as the cost function.
Perform the following steps for implementing gradient descent on the cost function.
1. **Define the Cost Function:**
   - Create a function `cost_function(x)` that returns the value of a quadratic function $f(x) = x^2 - 4x + 3$, which is the objective function we want to minimize.

2. **Define the Gradient:**
   - Write a function `derivative(x)` to compute the derivative of the cost function. The derivative is $f'(x) = 2x - 4$, which represents the slope at any point \( x \).

3. **Set Up Gradient Descent:**
   - Define a function `gradient_descent(learning_rate, initial_x, iterations)` to implement the gradient descent algorithm.
   - Initialize `x` with the starting value `initial_x`.

4. **Gradient Descent Loop:**
   - Use a `for` loop that runs for `iterations` number of times.
   - Inside the loop, calculate the gradient using the `derivative(x)` function.
   - Update the value of `x` by subtracting `learning_rate * gradient` to move \( x \) in the direction of minimizing the cost.
   - Calculate and print the current value of the cost function using `cost_function(x)` and display the iteration number.

5. **Return the Optimal Value:**
   - After the loop finishes, return the final value of \( x \), which is the estimated optimal point.

6. **Set Parameters:**
   - Set the `learning_rate`, `initial_x`, and `iterations` as needed for the gradient descent function.

7. **Run Gradient Descent:**
   - Call `gradient_descent()` with the appropriate parameters and print the final optimal value of \( x \) and the corresponding cost.

8. **Verify Result:**
   - Print the final `cost_function(optimal_x)` to confirm the cost at the optimal point.

```python
# Define the cost function
def cost_function(x):
  """returns costfunction of given input for 𝑥2−4𝑥+3 ."""
  return x**2-4*x+3

# Define the derivative of the cost function
def derivative(x):
  """ Return derivative of cost function 𝑥2−4𝑥+3 , which is 2x-4"""
  return 2*x-4

# Define the gradient descent function to minimize the cost function
def gradient_descent(learning_rate, initial_x, iterations):
    x = initial_x

    # Iterate for the given number of iterations
    for i in range(iterations):
        # Calculate the gradient of the cost function at current x
        grad_at_x=derivative(x)
        # Update x using gradient descent
        x=x-learning_rate*grad_at_x
        # Calculate the value of the cost function at the updated x
        cost_value=cost_function(x)
        # Print the current iteration, the value of x, and the value of the cost function
        print(f'current iteration: {i}, grad:{grad_at_x}, x:{x}, and the value of the cost function:{cost_value}')

    # Return the final value of x
    return x

# Parameters
learning_rate = 1.1  # Step size for gradient descent
initial_x = 0        # Initial value of x
iterations = 100      # Number of iterations

# Run gradient descent
optimal_x =gradient_descent(learning_rate,initial_x,iterations)

# Print the optimal value of x and the corresponding cost
print("Optimal x:", optimal_x)
print("Optimal cost:", cost_function(optimal_x))
```

Apply all 3 types of gradient descent on the breast cancer data available in `sklearn`.

### Steps for Implementing Gradient Descent with Logistic Regression:

1. **Load and Preprocess Dataset:**

2. **Add Bias Term:**

3. **Define Core Functions:**
   - **Sigmoid Function:** `sigmoid(z)` to compute the logistic function.
   - **Cost Function:** `compute_cost(X, y, theta)` to compute binary cross-entropy loss.
   - **Gradient Function:** `compute_gradient(X, y, theta)` to calculate the gradient.

4. **Implement Gradient Descent Methods:**
   - **Batch Gradient Descent:** `batch_gradient_descent()` updates weights using the entire dataset.
   - **Mini-Batch Gradient Descent:** `mini_batch_gradient_descent()` updates weights using small random batches.
   - **Stochastic Gradient Descent:** `stochastic_gradient_descent()` updates weights using a single random sample per iteration.

5. **Set Hyperparameters:**
   - Initialize weights `theta_initial` to zeros.
   - Set learning rate `alpha = 0.1`, iterations `1000`, and mini-batch size `32`.

6. **Run Gradient Descent:**
   - Execute all three methods (Batch, Mini-Batch, and Stochastic Gradient Descent) and track cost history.

   - Use the `predict()` function to make predictions on the test set with each model.

```python
# Import the required libraries
import numpy as np
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
from sklearn.metrics import classification_report

# Load the Breast Cancer dataset
cancer = load_breast_cancer()

# Split the features and target
X =cancer.data
y = cancer.target  # Binary classification (0 or 1)

# Standardize the dataset (important for gradient descent)
std_scaler=StandardScaler()

#sequence to follow . first split , then scale train, then add bias to them.
#1.If we scale on X before split it will cause data lekage
#2.If we scale after adding bias column, which is constant , If you run StandardScaler on a column of all 1s, the variance of that column is zero.
#Trying to scale it will result in dividing by zero, which turns your entire bias column into 0 or NaN

# Split the dataset into training and testing sets
X_train,X_test,y_train,y_test=train_test_split(X,y,random_state=42)
X_train=std_scaler.fit_transform(X_train) #perform fit_transform on training data
X_test=std_scaler.transform(X_test) #perform only transform on test data. No data lekage, it uses computed mean/std dev from training
# Add intercept term (bias) to X
X_train_b = np.c_[np.ones((X_train.shape[0], 1)), X_train]  # Add a column of ones for intercept
X_test_b = np.c_[np.ones((X_test.shape[0], 1)), X_test]     # Same for test set

# Sigmoid function for logistic regression
def sigmoid(z):
   # Clip z to prevent exp() overflow (bounds between -500 and 500)
    z = np.clip(z, -500, 500)
    return 1 / (1 + np.exp(-z))

# Logistic regression cost function (binary cross-entropy)
def compute_cost(X, y, theta):
    m = len(y) # number of samples
    h = sigmoid(X@theta) #X is m x d and theta is d X 1
    #clip h with very small vlaue to avoid log 0 and log 1
    eps=1e-15
    h=np.clip(h,eps,1-eps) #h will be between bourndries eps and 1-eps
    cost = -(1/m) * np.sum(y * np.log(h) + (1 - y) * np.log(1 - h))
    return cost

# Gradient of the cost function
#The Core Difference W^Tx is mathematical theory: Used in textbook equations for a single data point.
# X @theta is code implementation: Used in programming for a whole batch of data. (more than single data point)
def compute_gradient(X, y, theta):
    m = len(y)
    h = sigmoid(X@theta)
    gradient = (1/m) * (X.T @ (h - y))
    return gradient

# Batch Gradient Descent
def batch_gradient_descent(X, y, theta, alpha, iterations):
  cost_history=[]
  for i in range(iterations):
    cost = compute_cost(X, y, theta)
    cost_history.append(cost)
    grad=compute_gradient(X, y, theta) #compute grad on entire data set
    theta=theta-alpha*grad
    #print(f'Iteration:{i},Cost:{cost},grad:{grad}')


  return theta, cost_history

# Mini-Batch Gradient Descent
def mini_batch_gradient_descent(X, y, theta, alpha, iterations, batch_size):
    m = len(y)
    cost_history = []

    for i in range(iterations):
        # 1. Shuffle the data at the start of each epoch
        indices = np.random.permutation(m)
        X_shuffled = X[indices]
        y_shuffled = y[indices]

        # 2. Update theta iteratively over mini-batches
        for j in range(0, m, batch_size):
            X_mini_batch = X_shuffled[j:j+batch_size]
            y_mini_batch = y_shuffled[j:j+batch_size]

            grad = compute_gradient(X_mini_batch, y_mini_batch, theta)# compute grad on batch-size
            theta = theta - alpha * grad  # Updates dynamically inside the loop

        # 3. Compute and store total epoch cost AFTER updates
        epoch_cost = compute_cost(X, y, theta)
        cost_history.append(epoch_cost)

    return theta, cost_history

# Stochastic Gradient Descent.
def stochastic_gradient_descent(X, y, theta, alpha, iterations):
    m = len(y)
    cost_history = []

    for i in range(iterations):
        # In one epoch, we perform 'm' updates using a single random point each time
        for j in range(m):
            # 1. Pick a completely random sample index
            random_index = np.random.randint(0, m)

            # 2. Extract that single sample as a 2D row vector
            X_sample = X[random_index:random_index+1]
            y_sample = y[random_index:random_index+1]

            # 3. Compute gradient and update weights for this single point
            grad = compute_gradient(X_sample, y_sample, theta)
            theta = theta - alpha * grad

        # Track progress: Compute global cost once at the end of the epoch
        epoch_cost = compute_cost(X, y, theta)
        cost_history.append(epoch_cost)

    return theta, cost_history

# Prepare initial theta (weights)
n = X_train_b.shape[1]
theta_initial = np.zeros(n)

# Hyperparameters
alpha = 0.1
iterations = 1000
batch_size = 32  # For mini-batch

# ----------- Batch Gradient Descent -----------
# Find theta, cost history for batch gradient descent
theta_bgd,cost_history_bgd=batch_gradient_descent(X_train_b, y_train, theta_initial, alpha, iterations)

# ----------- Mini-Batch Gradient Descent -----------
# Find theta, cost history for mini batch gradient descent
theta_mbgd,cost_history_mbgd=mini_batch_gradient_descent(X_train_b, y_train,theta_initial, alpha, iterations, batch_size)

# ----------- Stochastic Gradient Descent -----------
# Find theta, cost history for batch gradient descent
theta_sgd,cost_history_sgd=stochastic_gradient_descent(X_train_b, y_train, theta_initial, alpha, iterations)

# Plot the cost history for each method
plt.figure(figsize=(6, 6))
plt.plot(range(iterations), cost_history_bgd, label='Batch Gradient Descent', color='r')
plt.plot(range(iterations), cost_history_mbgd, label='Mini-Batch Gradient Descent', color='g')
plt.plot(range(iterations), cost_history_sgd, label='Stochastic Gradient Descent', color='b')
plt.xlabel('Iterations')
plt.ylabel('Cost')
plt.title('Cost History for Different Gradient Descent Methods')
plt.legend()
plt.show()

# Predict function for logistic regression
# Create a helper function to turn raw matrix math into 0 or 1 predictions
def predict(X, theta):
    probabilities = sigmoid(X @ theta)
    return (probabilities >= 0.5).astype(int).ravel() # .ravel() ensures it matches y_test flat shape


# Evaluate all three models
y_pred_bgd =predict(X_test_b,theta_bgd)
y_pred_mbgd =predict(X_test_b,theta_mbgd)
y_pred_sgd =predict(X_test_b,theta_sgd)

# Print classification reports
print("Classification Report for Batch Gradient Descent:\n")
print(classification_report(y_test, y_pred_bgd))

print("\nClassification Report for Mini-Batch Gradient Descent:\n")
print(classification_report(y_test, y_pred_mbgd))

print("\nClassification Report for Stochastic Gradient Descent:\n")
print(classification_report(y_test, y_pred_sgd))
```

### Key Insights from Gradient Descent Optimization Methods

* **Rapid Convergence of SGD (Blue Line):**
  Stochastic Gradient Descent drops to a near-optimal cost almost instantly. While Batch GD only updates parameters once per epoch, SGD performs $m$ individual updates per epoch (where $m$ is the total dataset size). This higher frequency of updates drives rapid early-stage learning.

* **Stochastic Volatility / "Noisy" Updates (Blue Line):**
  The SGD curve exhibits a jagged, zig-zag pattern throughout all iterations. This fluctuation is caused by calculating the gradient using only one random data point at a time. Individual outliers pull the weights in slightly erratic directions, preventing a perfectly monotonic descent.

* **Stability of Batch Gradient Descent (Red Line):**
  The Batch GD curve is perfectly smooth because it computes the true, stable gradient over the entire dataset before making a move. However, because it only performs a single update per epoch, it requires significantly more iterations to reach the same low cost achieved by stochastic methods.

* **The Balanced Compromise of Mini-Batch GD (Green Line):**
  Mini-Batch GD provides the optimal middle ground. Averaging gradients over a moderate batch size (e.g., 32) dampens the noisy fluctuations seen in pure SGD. Simultaneously, updating weights multiple times per epoch allows it to converge much faster than Batch GD.
