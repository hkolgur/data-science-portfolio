##LeetCode DataSet
It contains the following info:

- **id**: problem id
- **title**: problem name
- **description**: problem description
- **is_premium**: whether the questions requires a premium account
- **difficulty**: easy, medium, or hard
- **solution_link**: how often the answer submitted is correct
- **acceptance_rate**: how often the answer submitted is correct
- **frequency**: how often the problem is attempted
- **url**: url to the problem
- **discuss_count**: how many comments are submitted by users
- **accepted**: how many times the answer was accepted
- **submissions**: how many times the answer was submitted
- **companies**: which companies were tagged as having asked this specific problem
- **related_topics**: related topics to the current problem
- **likes**: how many likes the problem got
- **dislikes**: how many dislikes the problem got
- **rating**: likes / (likes + dislikes)
- **asked_by_faang**: whether or not the question was asked by facebook, apple, amazon, google, or netflix
- **similar_questions**: similar problems with problem name, slug, and difficulty

### Question 7

#### Convert the datatypes for **accepted** and **submissions** columns

```python
#Solution 7(your code here)
# Convert 'accepted' and 'submissions' to numeric types

#check to see how many different abbreviations are used at end Like M for millions , k for 1000 etc
print(leetcode_df.submissions.str.slice(start=-1).unique())

#observaiton: we see only M and K at the end so we can expand them and convert to numeric

def parse_counts(val):
    converted_value=0
    if pd.isna(val):
      return 0
    elif  val[-1] in ('M', 'm') :
       return int(float(val[:-1]) *1000000)
    elif val[-1] in ('K', 'k'):
      return int(float(val[:-1]) *1000)
    try:
        return int(float(val))
    except ValueError:
        return 0



leetcode_df['submissions_converted']=leetcode_df['submissions'].apply(parse_counts)
leetcode_df['accepted_converted']=leetcode_df['accepted'].apply(parse_counts)
```

```python
# easy alternate for above logic:

leetcode_df['accepted_converted'] = (
    leetcode_df['accepted']
    .str.upper()
    .str.strip()
    .replace({'M': 'e6', 'K': 'e3'}, regex=True)
    .astype(float)
)

leetcode_df['submissions_converted'] = (
    leetcode_df['submissions']
    .str.upper()
    .str.strip()
    .replace({'M': 'e6', 'K': 'e3'}, regex=True)
    .astype(float)
)
```

```python
# Verify changes
print(leetcode_df.submissions_converted.dtype)
print(leetcode_df.accepted_converted.dtype)
```

### Question 10

```python
#Solution 10(your code here)
numeric_columns=leetcode_df.select_dtypes(include='number').columns.to_list()

q1=leetcode_df[numeric_columns].quantile(0.25)
q3=leetcode_df[numeric_columns].quantile(0.75)
iqr_df=q3-q1
def find_outliers(col):
  lower_bound= q1[col] - 1.5 * iqr_df[col]
  upper_bound= q3[col] + 1.5 * iqr_df[col]
  outlier_mask= (leetcode_df[col]<lower_bound) | (leetcode_df[col]>upper_bound)
  return leetcode_df[outlier_mask]

#we are creating a dictonary which as key as col and values as the list of outliers for this column
outliers= {col: find_outliers(col) for col in numeric_columns}

# Initialize an empty dictionary to hold the results
outliers_summary = {}

# Loop through each numeric column to isolate outliers cleanly
for col in numeric_columns:
    # Ensure there are actually outliers present for this specific column
    if not outliers[col].empty:
        # Correctly slice the ID and the target column from the main DataFrame
        outliers_summary[col] = outliers[col][['id', col]]

#outlier_summary=
# lower_bound= q1 -1.5 * iqr_df
# upper_bound= q3 + 1.5 * iqr_df

# outlier_mask= ((leetcode_df[numeric_columns]<lower_bound) | (leetcode_df[numeric_columns]>upper_bound )) .any(axis=1)
# outliers=leetcode_df[numeric_columns][outlier_mask]

```

### Question 11

#### Handle the outliers

```python
#Solution 11(your code here)
#Column-by-Column (Capping / Winsorization) . Instead of deleting entire row for viral columns.
# Because you can't drop cells individually, data scientists use the column-by-column approach to cap the outliers
# instead of deleting them. This is called Winsorization.Instead of throwing the row away, you change the extreme value to
# match your upper or lower bound.

# Create a copy of the dataframe for training
df_model = leetcode_df.copy()

for col in numeric_columns:
#     # Cap the high outliers at the upper bound
    df_model[col] = df_model[col].clip(lower=lower_bound[col], upper=upper_bound[col])
```

The average problem on the platform is moderately challenging, where roughly half of all submissions result in an accepted solution.

2. Right-Skewed Behavior (The Long Tail)Analysis: Look at how the right side of the graph slopes downward gradually all the way past 90%, while the left side cuts off sharply below 25%.Conclusion: This shows a right-skewed distribution. There is a wide variety of highly accessible, high-acceptance problems (mostly the "Easy" tier) stretching out the right side. On the flip side, truly brutal problems with an acceptance rate below 25% are exceptionally rare.

3. The Elite Few (The Tail): As you move right, the bars drop off drastically. Only a tiny fraction of elite problems possess a frequency score above 60. These are the famous, highly popular interview questions.

This specific shape tells us two critical things about how this column will behave during training:Log-Transformation Candidate: Models like Linear Regression, Logistic Regression, or Neural Networks perform poorly when a feature has a massive structural skew like this. If you plan to use this column as a predictive feature, you should consider a log transformation (np.log1p(df['frequency'])). This squashes the massive 0-20 spike and spreads the long tail out, turning it into a stable, bell-shaped normal distribution that models can learn from much more easily.

Tree-Based Model Readiness: If you plan to use tree-based algorithms like Random Forest or XGBoost, you can leave this column exactly as it is. Decision trees don't care about distribution shapes or skewness; they can split on these boundaries easily.
